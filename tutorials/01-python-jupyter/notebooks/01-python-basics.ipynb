## 1. Variables et types

Une variable est un nom qui désigne une valeur. Python devine le type tout seul.

```python
fs_hz = 116.2          # float : nombre à virgule (cadence mesurée, en hertz)
n_samples = 2315       # int : nombre entier
condition = "periodic" # str : texte
valid = True           # bool : vrai ou faux
print(type(fs_hz), type(n_samples), type(condition), type(valid))
duree_s = n_samples / fs_hz
print(f"{n_samples} mesures à {fs_hz} Hz durent {duree_s:.2f} s")
```

La ligne `f"... {duree_s:.2f} ..."` est une *f-string* : ce qui est entre accolades est remplacé par la valeur, ici avec 2 décimales.

## 2. Listes

Une liste contient plusieurs valeurs, dans l'ordre. **Le premier élément a l'indice 0.**

```python
conditions = ["stationary", "periodic", "impact"]
print(conditions[0])     # premier
print(conditions[-1])    # dernier
print(len(conditions))   # nombre d'éléments
conditions.append("inconnu")
print(conditions)
```

### Le *slicing* : extraire une tranche

`liste[debut:fin]` prend les éléments de `debut` (inclus) à `fin` (**exclu**).

```python
mesures = [10, 11, 12, 13, 14, 15, 16, 17]
print(mesures[0:3])   # les 3 premières : indices 0, 1, 2
print(mesures[3:])    # à partir de l'indice 3
print(mesures[::2])   # une sur deux
```

**Lien avec le cours :** une fenêtre du TP01, ce sont les mesures `[debut : debut + 256]`.

## 3. Boucles et conditions

```python
for i, c in enumerate(conditions):
    if c == "inconnu":
        print(i, c, "-> pas une classe du modèle")
    else:
        print(i, c)
```

## 4. Fonctions

Une fonction regroupe un calcul réutilisable. Les lignes entre triples guillemets expliquent ce qu'elle fait.

```python
def duree(n, fs):
    """Durée en secondes de n mesures prises à la cadence fs (Hz)."""
    return n / fs

print(duree(256, 116.2))   # une fenêtre du TP01
print(duree(100, 100.0))   # 100 mesures à 100 Hz
```

## 5. Dictionnaires

Un dictionnaire associe une clé à une valeur. C'est le format des fichiers JSON du cours (`labels.json`, `preprocessing.json`…).

```python
qc = {"n_samples": 2315, "fs_hz": 116.24, "lacunes": 0}
print(qc["fs_hz"])
for cle, valeur in qc.items():
    print(f"{cle:10s} = {valeur}")
```
