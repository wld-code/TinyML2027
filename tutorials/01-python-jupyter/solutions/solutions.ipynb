## Exercice 1 — durée d'une fenêtre

```python
def duree(n, fs):
    return n / fs
for fs in (116.2, 100.0):
    print(f"256 mesures à {fs} Hz : {duree(256, fs):.3f} s")
```

## Exercice 2 — RMS d'une liste, sans NumPy puis avec

```python
import math
import numpy as np
valeurs = [0.1, -0.2, 0.15, -0.05]
rms_boucle = math.sqrt(sum(v * v for v in valeurs) / len(valeurs))
rms_numpy = float(np.sqrt(np.mean(np.array(valeurs) ** 2)))
print(rms_boucle, rms_numpy)
assert abs(rms_boucle - rms_numpy) < 1e-12
```

## Exercice 3 — trouver les défauts du fichier (notebook 03, partie 4)

```python
import pandas as pd
df = pd.read_csv("../data/sample_acceleration.csv")
dseq = np.diff(df.seq.to_numpy())
dtu = np.diff(df.t_us.to_numpy())
print("seq non consécutifs après les lignes :", np.flatnonzero(dseq != 1).tolist())
print(df.iloc[np.flatnonzero(dseq != 1)[0] - 1: np.flatnonzero(dseq != 1)[0] + 3])
print("t_us non croissants après les lignes :", np.flatnonzero(dtu <= 0).tolist())
i = int(np.flatnonzero(dtu <= 0)[0])
print(df.iloc[i - 1:i + 3])
```

**Lecture :** il manque la mesure `seq = 400` (on passe de 399 à 401), et les instants des mesures 700 et 701 sont inversés. Il ne faut ni interpoler la mesure manquante ni réordonner les lignes en silence : on exclut les fenêtres concernées et on note la raison.
